```python
import pandas as pd
import re

bike_data = []

bike_headings = soup.find_all("h3", class_="o-c6")

for heading in bike_headings:

    # Find the bike listing container
    link = heading.find_parent("a", href=True)
    if not link:
        continue

    card = link.parent.parent

    # Bike name and brand
    bike_name = heading.get_text(" ", strip=True)
    brand = bike_name.split()[0]

    # Get text from the listing
    text = card.get_text(" ", strip=True)

    # Extract rating
    rating_match = re.search(r'(\d+(?:\.\d+)?)\s*/\s*5', text)
    rating = float(rating_match.group(1)) if rating_match else None

    # Extract review count
    reviews_match = re.search(r'([\d,]+)\s+Ratings?', text)
    review_count = (
        int(reviews_match.group(1).replace(",", ""))
        if reviews_match else None
    )

    # Extract engine capacity
    engine_match = re.search(r'([\d.]+)\s*cc', text)
    engine_cc = float(engine_match.group(1)) if engine_match else None

    # Extract mileage
    mileage_match = re.search(r'([\d.]+)\s*kmpl', text)
    mileage = float(mileage_match.group(1)) if mileage_match else None

    # Extract power
    power_match = re.search(r'([\d.]+)\s*bhp', text)
    power_bhp = float(power_match.group(1)) if power_match else None

    # Extract weight
    weight_match = re.search(r'([\d.]+)\s*kg', text)
    weight_kg = float(weight_match.group(1)) if weight_match else None

    # Extract price
    price_match = re.search(r'₹\s*([\d,]+)', text)
    price = (
        int(price_match.group(1).replace(",", ""))
        if price_match else None
    )

    bike_data.append({
        "bike_name": bike_name,
        "brand": brand,
        "price": price,
        "mileage": mileage,
        "engine_cc": engine_cc,
        "power_bhp": power_bhp,
        "rating": rating,
        "weight_kg": weight_kg,
        "review_count": review_count,
        "product_url": link["href"]
    })

df = pd.DataFrame(bike_data)

print("Dataset shape:", df.shape)
print(df.head(10).to_string(index=False))
```

```python
#Hypothesis Testing — Price and Mileage
from scipy import stats
r, p_value = stats.pearsonr(df["price"], df["mileage"])

print("Correlation:", round(r, 3))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("There is a statistically significant linear relationship.")
else:
    print("There is not enough evidence of a significant linear relationship.")
```

#### ANOVA Test — Compare Bike Prices Across Brands

```python
from scipy.stats import f_oneway

# Group prices by brand
brand_groups = [
    group["price"].values
    for brand, group in df.groupby("brand")
    if len(group) >= 2
]

# Perform ANOVA test
f_stat, p_value = f_oneway(*brand_groups)

print("F-statistic:", round(f_stat, 3))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("Average prices differ significantly across brands.")
else:
    print("No significant difference found in average prices.")
```
